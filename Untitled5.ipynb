```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# 1. Dataset
X = np.array([[1], [2], [3], [4], [5], [6]])  # Study hours
y = np.array([0, 0, 0, 1, 1, 1])  # 0 = Fail, 1 = Pass

# 2. Train Logistic Regression
model = LogisticRegression()
model.fit(X, y)

w = model.coef_[0][0]
b = model.intercept_[0]
print(f"Logistic Regression function: f(x) = sigmoid({w:.4f} * x + {b:.4f})")

# 3. Generate points for smooth sigmoid curve
x_range = np.linspace(0, 8, 200)
z = w * x_range + b
sigmoid = 1 / (1 + np.exp(-z))

# 4. Plot
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color="blue", label="Actual Data (Pass/Fail)", zorder=3)
plt.plot(x_range, sigmoid, color="red", label="Sigmoid Curve", linewidth=2)
plt.axhline(
    0.5,
    color="gray",
    linestyle="--",
    alpha=0.7,
    label="Decision Boundary (P = 0.5)",
)
plt.axvline(
    -b / w,
    color="green",
    linestyle=":",
    alpha=0.7,
    label=f"Threshold (x = {-b/w:.2f})",
)

plt.xlabel("Study Hours (x)", fontsize=12)
plt.ylabel("Probability of Passing", fontsize=12)
plt.title("Logistic Regression Sigmoid Curve", fontsize=14)
plt.ylim(-0.05, 1.05)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

# 5. Predict for x = 7
x_new = np.array([[7]])
y_pred_prob = model.predict_proba(x_new)[0][1]
y_pred = model.predict(x_new)[0]

print(f"\nInput x = 7")
print(f"Probability of passing: {y_pred_prob:.4f}")
print(f"Predicted label: {y_pred}")
```
